CyclicRotation

```python
A = [3,8,9,7,6]
K = int(input("K= "))

def solution(A,K):
  N = len(A)

  if N == 0:
    return A

  if 0 <= K < 100:
    for i in range(K):
      last = A[N-1]
      for j in range(N-1,0,-1):
        A[j] = A[j-1]

      A[0] = last
    return A


print(solution(A,K))
```

FrogJmp

```python
def solution(X,Y,D):
  distance = Y - X

  jump = distance//D

  if distance % D != 0:
    jump = jump + 1

  return jump

jumps = solution(10,85,30)
print(jumps)
```

PermCheck

```python
def solution(A):
  N = len(A)

  for i in range(1, N+1):
    count = 0
    for j in range(N):
      if A[j] == i:
        count = count + 1


    if count != 1:
      return 0


  return 1
print(solution([4,1,3,2]))
print(solution([4,1,3]))

```

FrogRiverOne

```python

def solution(X,A):

  leaves = []

  for i in range(X):
    leaves.append(0)

  for K in range(len(A)):
    position = A[K]

    if leaves [position - 1] == 0:
        leaves[position - 1] = 1

    if 0 not in leaves:
      return K

print(solution(5,[1,3,1,4,2,3,5] ))
```

PassingCars

```python
def solution(A):
  east = 0
  pairs = 0

  for i in range(len(A)):
    if A[i] == 0:
      east = east + 1

    elif A[i] == 1:
      pairs = pairs + east

    if pairs > 1000000000:
      return -1

  return pairs

print(solution([0, 1, 0, 1, 1]))
```
